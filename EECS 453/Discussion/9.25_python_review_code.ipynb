```python
# define class
class Foo:

    # optinal constructor
    def __init__(self, x):
        # first parameter "self" for instance reference, like "this" in JAVA
        self.x = x

    # instance method
    def printX(self): # instance reference is required for all function parameters
        print(self.x)

    # class methods, most likely you will never need this
    @classmethod
    def printHello(self):
        print("hello")

obj = Foo(6)
obj.printX()
Foo.printHello()
```

### 3. Iterables

```python
alist = list()  # linear, size not fixed, not hashable
print(len(alist)) # gets size of any iterable type
atuple = tuple() # linear, fixed size, hashable if all elements in tuple are hashable
not_hashable_tuple = ([1, 2], 10)
print(type(not_hashable_tuple))
hash(not_hashable_tuple)

## Hashable means Python can compute a stable hash value for an object.
## That lets the object be used as a dictionary key or a set element.
## Strings and numbers are hashable; lists and dictionaries are not.
```

```python
adict = dict()  # hash table, not hashable, stores (key,value) pairs
aset = set([1,1,2,3])    # hash table, like dict but only stores keys
print(aset)
```

```python
original = [[1, 2]]
shallow = original.copy()

shallow.append([3, 4])    # edit the outer part: original unchanged
print(f'shallow: {shallow}')
print(f'original: {original}')

shallow[0].append(99)      # edit shared inner part list：original will change
print(f'shallow: {shallow}')
print(f'original: {original}')
```

```python
# examplar tuple usage
# creating a dictionary to store ngram counts
d = dict()
d[("a","cat")] = 10
d["dog"] = 14

print(d)
```

```python
"""
List: not hashable (i.e. can't use as dictionary key)
      dynamic size
      allows duplicates and inconsistent element types
      dynamic array implementation
"""
# list creation
alist = []          # empty list, equivalent to list()
alist = [1,2,3,4,5] # initialized list

print(alist[0])
alist[0] = 5
print(alist)
```

```python
print("-"*10)
print([5]*10)
# list indexing
print(alist[0]) # get first element (at index 0)
print(alist[-2]) # get last element (at index len-1)
print(alist[3:]) # get elements starting from index 3 (inclusive)
print(alist[:3]) # get elements stopping at index 3 (exclusive)
print(alist[2:4]) # get elements within index range [2,4)
print(alist[6:]) # prints nothing because index is out of range
print(alist[::-1]) # returns a reversed list
```

```python
print("-"*10)
# list modification
alist.append("new item") # insert at end
alist.insert(0, "new item") # insert at index 0
print(alist)
alist.extend([2,3,4]) # concatenate lists
print(alist)
# above line is equivalent to alist += [2,3,4]
alist.index("new item") # search by content
alist.remove("new item") # remove by content
alist.pop(0) # remove by index
print(alist)
```

```python
print("-"*10)
if "new item" in alist:
    print("found")
else:
    print("not found")

print("-"*10)
# list traversal
for ele in alist:
    print(ele)

print("-"*10)
# or traverse with index
for i, ele in enumerate(alist):
    print(i, ele)
```

```python
"""
Tuple: hashable (i.e. can use as dictionary key)
       fixed size (no insertion or deletion)
"""
# it does not make sense to create empty tuples
atuple = (1,2,3,4,5)
 # or you can cast other iterables to tuple
atuple = tuple([1,2,3])
print(atuple)
# indexing and traversal are same as list
```

```python
"""
Dict: not hashable
      dynamic size
      no duplicates allowed
      hash table implementation which is fast for searching
"""
# dict creation
adict = {} # empty dict, equivalent to dict()
adict = {'a':1, 'b':2, 'c':3}
print(adict)

# get all keys in dictionary
print(adict.keys())

# get value paired with key
print(adict['a'])
key = 'e'

# NOTE: accessing keys not in the dictionary leads to exception
if key in adict:
    print(adict[key])

# add or modify dictionary entries
adict['e'] = 10 # insert new key
adict['e'] = 5  # modify existing keys

print("-"*10)
# traverse keys only
for key in adict:
    print(key, adict[key])

print("-"*10)
# or traverse key-value pairs together
for key, value in adict.items():
    print(key, value)

print("-"*10)
# NOTE: Checking if a key exists
key = 'e'
if key in adict: # NO .keys() here please!
    print(adict[key])
else:
    print("Not found!")
```

```python
"""
Special dictionaries
"""
# set is a dictionary without values
aset = set()
aset.add('a')
print(aset)

# deduplication short-cut using set
alist = [1,2,3,3,3,4,3]
alist = list(set(alist))
print(alist)

# default_dictionary returns a value computed from a default function
#     for non-existent entries
from collections import defaultdict
adict = defaultdict(lambda: 'unknown')
adict['cat'] = 'feline'
print(adict['cat'])
print(adict['dog'])
```

```python
# counter is a dictionary with default value of 0
#     and provides handy iterable counting tools
from collections import Counter

# initialize and modify empty counter
counter1 = Counter()
counter1['t'] = 10
counter1['t'] += 1
counter1['e'] += 1
print(counter1)
print("-"*10)

# initialize counter from iterable
counter2 = Counter("letters to be counted")
print(counter2)
print("-"*10)

# computations using counters
print("sum: ", counter1 + counter2)
print("difference: ", counter2 - counter1)
print("intersection: ", counter1 or counter2) # or for intersection, and for union
```

```python
c = np.array([[1, 2], [3, 4]])

# Compute the inverse of c. Requires c to be square and invertible.
print(np.linalg.inv(c))

b = np.array([1, 1])

# Solve c @ x = b by explicitly computing the inverse.
# This works here, but is generally not the preferred approach.
print(np.linalg.inv(c) @ b)

# Solve the linear system c @ x = b directly.
# Preferred: usually faster and more numerically stable than inv(c) @ b.
print(np.linalg.solve(c, b))
```

```python
# vector dot product
v1 = np.array([1,2]) # shape (2,)
v2 = np.array([3,4]) # shape (2,)
print(v1.dot(v2))
print(np.dot(v1,v2))
print(v1@v2)
```

```python
# Compute the outer product: result[i, j] = v1[i] * v2[j].
# If v1 has length m and v2 has length n, the result has shape (m, n).
print(np.outer(v1, v2))

# Equivalent method: reshape v1 into a column vector (m, 1)
# and v2 into a row vector (1, n), then multiply the two matrices.
print(v1.reshape(-1, 1).dot(v2.reshape(1, -1)))
```

```python
# Matrix-vector multiplication: compute m @ v1.
# m has shape (2, 2), so v1 must have shape (2,).
# The result is a 1D array with shape (2,).

print(m @ v1)             # @ operator: standard, readable notation
print(m.dot(v1))          # dot(): gives the same result for a 2D matrix and 1D vector
print(np.matmul(m, v1))   # matmul(): gives the same result here; also handles batches

# For a 2D matrix multiplied by a 1D vector, all three do the same thing.
# @ is the operator, np.matmul is the corresponding function, and dot is an older function with different rules for higher-dimensional arrays.
# For ordinary matrix multiplication, @ is preferred.
```

```python
# matrix multiplication
A = np.ones((4,3)) # 4,3
B = np.ones((3,2)) # 3,2 --> 4,2
print(A @ B)       # same as A.dot(B)
```

```python
# broadcasting
c = np.ones([4,4])
# automatic repetition along axis
d = np.array([1,2,3,4]).reshape(4,1)
print(c.shape)
print(d.shape)
print(c + d)
```

```python
# Generate 15 samples, each with 5 features. Shape: (15, 5).
samples = np.random.random((15, 5))

# Insert a size-1 dimension in the middle: (15, 1, 5).
# Each sample is now positioned to pair with every sample in the other array.
print(samples[:, np.newaxis, :].shape)

# Insert a size-1 dimension at the front: (1, 15, 5).
# np.newaxis is equivalent to None; it adds an axis without copying the data.
print(samples[np.newaxis, :, :].shape)

# Broadcasting compares every pair of samples:
# (15, 1, 5) - (1, 15, 5) -> (15, 15, 5).
# diff[i, j, :] contains the feature-wise difference samples[i] - samples[j].
diff = samples[:, np.newaxis, :] - samples[np.newaxis, :, :]
print(diff.shape)

# Take the Euclidean norm across the 5 features to get pairwise distances.
# distances[i, j] is the distance between samples[i] and samples[j].
distances = np.linalg.norm(diff, axis=2)  # Shape: (15, 15).
```

```python
# speed test: numpy vs list
a = np.ones((100,100))
b = np.ones((100,100))

def matrix_multiplication(X, Y):
    result = [[0]*len(Y[0]) for _ in range(len(X))]
    for i in range(len(X)):
        for j in range(len(Y[0])):
            for k in range(len(Y)):
                result[i][j] += X[i][k] * Y[k][j]
    return result

import time

# run numpy matrix multiplication for 10 times
start = time.time()
for _ in range(10):
    a @ b
end = time.time()
print("numpy spends {} seconds".format(end-start))

# run list matrix multiplication for 10 times
start = time.time()
for _ in range(10):
    matrix_multiplication(a,b)
end = time.time()
print("list operation spends {} seconds".format(end-start))

# the difference gets more significant as matrices grow in size!
```

```python
# Generate x values from 0 (inclusive) to 2 (exclusive) in steps of 0.01.
# This gives 200 values: 0.00, 0.01, ..., 1.99.
x = np.arange(0, 2, 0.01)

# Compute y = 1 + sin(2πx) for each x value.
# The sine wave has a period of 1 and is shifted up by 1.
y = 1 + np.sin(2 * np.pi * x)

# Plot the 200 (x, y) points and connect them with a line.
# Axis tick labels are just markers; they do not show every x value.
plt.plot(x, y)
```
